```python
# Code Example
import tensorflow
import keras
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from keras.layers import Dense
from keras import Sequential
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
```

```python
# Load Dataset
df=pd.read_csv('ushape.csv', header=None)
```

```python
# Give column name
df.columns=['X', 'Y', 'class']
```

```python
# Check the data
df.head(3)
```

```python
# Scatter plot
plt.scatter(df['X'], df['Y'], c=df['class'])
```

```python
# Extract input and output
X=df.iloc[:, 0:2].values
y=df.iloc[:, -1].values
```

```python
# Create neural network
model=Sequential()
model.add(Dense(10, activation='relu', input_dim=2))
model.add(Dense(1, activation='sigmoid'))
```

```python
# Structure of the model
model.summary()
```

```python
model.get_weights()
```

# Initialize all weights as Zero

```python
initial_wt=model.get_weights()
```

```python
# Initialize wight as zero
initial_wt[0] = np.zeros(model.get_weights()[0].shape)
initial_wt[1] = np.zeros(model.get_weights()[1].shape)
initial_wt[2] = np.zeros(model.get_weights()[2].shape)
initial_wt[3] = np.zeros(model.get_weights()[3].shape)
```

```python
model.set_weights(initial_wt)
```

```python
model.get_weights()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history = model.fit(X,y,epochs=100,validation_split=0.2)
```

```python
model.get_weights()
```
