```python
# Code Example
import tensorflow
import keras
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from keras.layers import Dense
from keras import Sequential
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
```

```python
# Load Dataset
df=pd.read_csv('ushape.csv', header=None)
```

```python
# Give column name
df.columns=['X', 'Y', 'class']
```

```python
# Scatter plot
plt.scatter(df['X'], df['Y'], c=df['class'])
```

```python
X = df.iloc[:,0:2].values
y = df.iloc[:,-1].values
```

```python
model=Sequential()
model.add(Dense(10, activation='tanh', input_dim=2))
model.add(Dense(10, activation='tanh'))
model.add(Dense(10, activation='tanh'))
model.add(Dense(10, activation='tanh'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.summary()
```

```python
model.get_weights()
```

```python
initail_wt=model.get_weights()
```

```python
initail_wt[0]=np.random.randn(2,10)*np.sqrt(1/2)
initail_wt[1]=np.zeros(model.get_weights()[1].shape)
initail_wt[2]=np.random.randn(10,10)*np.sqrt(1/10)
initail_wt[3]=np.zeros(model.get_weights()[3].shape)
initail_wt[4]=np.random.randn(10,10)*np.sqrt(1/10)
initail_wt[5]=np.zeros(model.get_weights()[5].shape)
initail_wt[6]=np.random.randn(10,10)*np.sqrt(1/10)
initail_wt[7]=np.zeros(model.get_weights()[7].shape)
initail_wt[8]=np.random.randn(10,1)*np.sqrt(1/10)
initail_wt[9]=np.zeros(model.get_weights()[9].shape)
```

```python
model.set_weights(initail_wt)
```

```python
model.get_weights()
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(X,y,epochs=100, validation_split=0.2)
```
