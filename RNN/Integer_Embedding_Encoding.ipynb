```python
# There are two way to convert text to vector:
    # Integer Encoding
    # Embeddings
# Integer encoding in an RNN converts each word/token into a unique integer ID
# A text sequence such as "I love deep learning" to be represented as [1, 2, 3, 4]
```

```python
# A tokenizer converts raw text into smaller units called tokens, such as words or subwords
# Tokenization → breaks text into tokens.
# Integer encoding → converts those tokens into numbers.
```

```python
# Tokenizer
    # Breaks text into tokens.
# Integer encoding
    # Maps tokens to integer IDs.
# Embedding
    # Converts integer IDs into learned vectors.
```

```python
# Code Example
```

```python
# Import libraries
import numpy as np
import tensorflow
import keras
from keras import Sequential
from keras.layers import Embedding
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.utils import pad_sequences
```

```python
# Create a doc file
docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'modi ji ki jai',
		'inquilab zindabad']
```

```python
# Preprocess the doc file
```

```python
# Create tokenizer object
tokenizer = Tokenizer()
```

```python
# Fit on the docs
tokenizer.fit_on_texts(docs)
```

```python
# Check word index
vocab_size = len(tokenizer.word_index) + 1
```

```python
# Counts how many times each word occurs
tokenizer.word_counts
len(tokenizer.word_counts)
```

```python
# how many text documents/sentences were given to the tokenizer
tokenizer.document_count
```

```python
# converts text documents into sequences of integer IDs
sequences=tokenizer.texts_to_sequences(docs)
sequences
```

```python
# pad_sequences() is used to make all text sequences the same length.
# This is very important before giving sequences to an RNN/LSTM/GRU
# neural networks usually expect inputs with the same shape.

sequences = pad_sequences(sequences,padding='post')  # Add 0 at last seqence
sequences
```

```python
# Perform Embedding

# Add embedding layer
model = Sequential()

model.add(
    Embedding(
        input_dim=vocab_size,
        output_dim=2
    )
)

model.build(input_shape=(None, sequences.shape[1]))

model.summary()
```

```python
model.compile('adam','accuracy')
```

```python
pred = model.predict(sequences)
print(pred)
```
